### Ejercicio 0

Importa los paquetes y módulos que necesites a lo largo del notebook

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

%matplotlib inline
plt.rcParams['figure.figsize'] = (16, 9)
plt.style.use('ggplot')

from sklearn.decomposition import PCA

from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
```

### Ejercicio 1: Importa los datos
1. Importa los datos: *nutrition.csv*
2. Observa las columnas que hay, así como su tipo.

```python
df = pd.read_csv("./data/nutrition.csv")
df.head()
```

### Ejercicio 2


Analiza la correlación entre las variables numéricas. 

```python
# Seleccionar solo columnas numéricas
df_num = df.select_dtypes(include="number").drop(columns="ID")

# Matriz de correlación
corr_matrix = df_num.corr()

# Heatmap
plt.figure(figsize=(18, 14))
sns.heatmap(
    corr_matrix,
    cmap="coolwarm",
    center=0,
    annot=False,
    linewidths=0.3
)
plt.title("Matriz de correlación - Variables numéricas", fontsize=14)
plt.tight_layout()
plt.show()
```

### Ejercicio 3

¿Qué pasa con las columnas USRDA? ¿Hay que tomar alguna decisión?

Son los mismos nutrientes expresados como % de la dosis diaria recomendada, por lo que no aportan infomación nuemva

```python
# Eliminar todas las columnas USRDA
usrda_cols = [col for col in df.columns if col.endswith("_USRDA")]
print("Columnas eliminadas:", len(usrda_cols))

df = df.drop(columns=usrda_cols)
print("Shape resultante:", df.shape)
```

### Ejercicio 4: Variables Categóricas

Podríamos analizar las variables no numéricas del dataset e intentar obtener más información, pero por simplicidad en este caso procedemos a eliminarlas. Elimina las variables de texto del dataset.

```python
cat_cols = df.select_dtypes(include="object").columns.tolist()
print("Columnas eliminadas:", cat_cols)

df = df.drop(columns=cat_cols)
print("Shape resultaante:", df.shape)
```

### Ejercicio 5: Distribuciones

Analiza la distribución de las variables/columnas/features restantes. Recuerda que PCA es altamente sensible a las escala, así que transforma aquellas variables que creas necesario antes de proceder a escalarlas.


```python
df.hist(figsize=(20, 15), bins=50, edgecolor='black')
plt.suptitle("Distribución de variables numéricas", fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Eliminar ID antes de transformar
df = df.drop(columns="ID")

# Aplicar log1p a todas las columnas (todas tienen sesgo derecho)
df_log = np.log1p(df)

# Verificar resultado
df_log.hist(figsize=(20, 15), bins=50, edgecolor='black')
plt.suptitle("Distribución tras log1p", fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```

### Ejercicio 6: Estandarizado
Estandariza cada variable. (No es necesario que dividas en train y test por simplificar).

```python
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_log)

# Convertir en DataFrame para mantener los nombred de las columnas
df_scaled = pd.DataFrame(df_scaled, columns=df_log.columns)

print("Shape:", df_scaled.shape)
df_scaled.head()
```

### Ejercicio 7: Implementación del PCA
Prueba a montar un PCA con todos los componentes. Para ello utiliza los datos previamente transformados y escalados.

```python
pca = PCA()
pca.fit(df_scaled)

expl_variance = pca.explained_variance_ratio_
cumul_variance = np.cumsum(expl_variance)

print(f"Número de componentes: {pca.n_components_}")
print(f"Varianza explicada por PC1: {expl_variance[0]:.2%}")
print(f"Varianza explicada por PC2: {expl_variance[1]:.2%}")
print(f"Varianza acumulada PC1+PC2: {cumul_variance[1]:.2%}")
```

### Ejercicio 8

Muestra en un dataframe la relación de cada componente con cada feature. En general, este paso no lo haremos salvo que sea necesario a posteriori explicar la influencia de alguna componente en función de las features originales, pero ahora, muéstralas.

```python
# Loadings: relación de cada componente con cada feature
loadings = pd.DataFrame(
    pca.components_,
    columns=df_scaled.columns,
    index=[f"PC{i+1}" for i in range(pca.n_components_)]
)

loadings
```

### Ejercicio 9: Aportación de cada PCA
Visualiza en un diagrama de líneas la suma acumulada de la varianza explicativa del PCA.

Si tuviéses que quedarte con 70-75 % de la varianza original, ¿con cuántos Principal Components te quedarías?

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

componentes = range(1, len(expl_variance) + 1)

# Varianza explicada por componente
axes[0].bar(componentes, expl_variance, color="steelblue")
axes[0].set_xticks(list(componentes))
axes[0].set_xlabel("Componente principal")
axes[0].set_ylabel("Varianza explicada")
axes[0].set_title("Varianza explicada por componente")

# Varianza acumulada
axes[1].plot(componentes, cumul_variance, marker="o", color="steelblue")
axes[1].axhline(y=0.70, color="red", linestyle="--", label="70%")
axes[1].axhline(y=0.75, color="orange", linestyle="--", label="75%")

axes[1].set_xticks(list(componentes))
axes[1].set_xlabel("Número de componentes")
axes[1].set_ylabel("Varianza acumulada")
axes[1].set_title("Varianza explicada acumulada")
axes[1].legend()

plt.tight_layout()
plt.show()
```

Me quedaría con 6 componentes

### Ejercicio 10: Reducción de features

Crea un nuevo PCA con X Principal Components, siendo X la cantidad de PC escogidos en el apartado anterior.

Obtén el nuevo dataset con el mismo número de registros que el original, pero en este caso con X features, que representan los PC elegidos.

```python
# PCA con 6 componentes
pca_6 = PCA(6)
df_pca_6 = pca_6.fit_transform(df_scaled)

# Convertir a DataFrame
df_pca_6 = pd.DataFrame(
    df_pca_6,
    columns=[f"PC{i+1}" for i in range(6)]
)

print(f"Varianza explicada acumulada con 6 componentes: {pca_6.explained_variance_ratio_.sum():.2%}")
print(f"Shape resultante: {df_pca_6.shape}")
df_pca_6.head()
```
